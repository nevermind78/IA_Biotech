## 3. Les chaînes de caractères : une séquence d'ADN est un texte

```python
adn = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"

print("Longueur :", len(adn))
print("3 premières bases (codon start) :", adn[:3])       # découpage [début:fin]
print("3 dernières bases :", adn[-3:])
print("Nombre de G :", adn.count("G"))
print("En minuscules :", adn.lower())
```

### ✍️ Exercice 1 — Taux de GC
Le **taux de GC** (proportion de G + C) influence la stabilité de l'ADN. Calculez-le pour la
séquence `adn` (résultat attendu : environ 0,54).

```python
# ✍️ À vous : remplacez None
taux_gc = None

print("Taux de GC :", taux_gc)
```

```python
#@title 💡 Solution
taux_gc = (adn.count("G") + adn.count("C")) / len(adn)
print(f"Taux de GC : {taux_gc:.2f}")
```

## 5. Dictionnaires : associer une clé à une valeur

```python
code_genetique = {"ATG": "M", "GCC": "A", "ATT": "I", "GTA": "V", "GGC": "G", "CGC": "R",
                  "TGA": "*", "AAG": "K", "GGT": "G", "CCC": "P", "GAT": "D", "AGA": "R", "TAG": "*"}
print("ATG code pour :", code_genetique["ATG"])
print("Codon inconnu → valeur par défaut :", code_genetique.get("NNN", "?"))
```

## 6. Fonctions : réutiliser du code

```python
def traduire(sequence_adn):
    """Traduit une séquence d'ADN en protéine (s'arrête au premier codon stop)."""
    proteine = ""
    for i in range(0, len(sequence_adn) - 2, 3):
        aa = code_genetique.get(sequence_adn[i:i + 3], "?")
        if aa == "*":
            break
        proteine += aa
    return proteine

print("Protéine :", traduire(adn))
```

*Indices :* un dictionnaire `{"A": "T", ...}` et `seq[::-1]` pour inverser.

```python
# ✍️ À vous
def complement_inverse(seq):
    return None

print(complement_inverse("ATGC"))
```

```python
#@title 💡 Solution
def complement_inverse(seq):
    paires = {"A": "T", "T": "A", "C": "G", "G": "C"}
    return "".join(paires[b] for b in seq)[::-1]

print(complement_inverse("ATGC"))
```

## 7. Pandas : le tableur du data scientist

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "gene": ["BRCA1", "TP53", "EGFR", "KRAS", "MYC", "GAPDH"],
    "chromosome": ["17", "17", "7", "12", "8", "12"],
    "expression_tumeur": [2.1, 0.4, 8.7, 5.2, 9.1, 10.0],
    "expression_sain": [3.0, 3.5, 2.2, 2.0, 2.5, 10.2],
})
df
```

```python
print("Dimensions :", df.shape)
df["log2FC"] = np.log2(df["expression_tumeur"] / df["expression_sain"]).round(2)   # nouvelle colonne calculée
display(df)

print("Gènes surexprimés dans la tumeur (log2FC > 1) :")
display(df[df["log2FC"] > 1])                          # filtrer des lignes

print("Moyenne d'expression par chromosome :")
display(df.groupby("chromosome")[["expression_tumeur", "expression_sain"]].mean())
```

## 8. Un graphique en 3 lignes

```python
import matplotlib.pyplot as plt

df.plot(x="gene", y=["expression_sain", "expression_tumeur"], kind="bar", figsize=(7, 3.5))
plt.ylabel("Expression (u.a.)"); plt.title("Expression des gènes : tissu sain vs tumeur")
plt.xticks(rotation=0); plt.show()
```
